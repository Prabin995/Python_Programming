```python
has_lottery_word = True
unknown_sender = True
has_link = True
sender_is_bank = False

if sender_is_bank:
  print("Email is NOT SPAM")

else:
  negativenum=0
  if has_lottery_word:
    negativenum=negativenum+3
  if unknown_sender:
    negativenum=negativenum+1
  if has_link:
    negativenum=negativenum+1

  if negativenum>=3:
    print("Email is SPAM")

  else:
    print("Email is NOT SPAM")




```

```python
scores = [3, 1, 4, 1, 5]
length=len(scores)
toltaladd=0
for x in scores:
  toltaladd=toltaladd+x
result=toltaladd/length
print(f"Total score is {toltaladd} and average is {result}.")
```

```python
scores = [5, 1, 3, 0, 4, 2, 3, 1]
spam=0
nonspam=0

for x in scores:
  if x>=3:
    spam=spam+1
  else:
    nonspam=nonspam+1

print(f"{spam} emails are spam and {nonspam} emails are non spam.")
```

```python
scores = [5, 1, 3, 0, 4, 2, 3, 1]
biggest = scores[0]

for x in scores:
    if x > biggest:
        biggest = x

print(biggest)
```

```python
scores = [-5, -2, -9]
smallest = scores[0]

for x in scores:
    if x < smallest:
        smallest = x

print(smallest)
```
