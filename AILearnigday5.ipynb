```python
my_list = [3, 7, 7, 2, 9]
S_list = sorted(my_list)
print("Sorted list:", S_list)

# Calculate mean
total = sum(S_list)
mean_f = total / len(S_list)
print("Mean:", mean_f)

# Calculate variance and standard deviation
variance_sum = sum((x - mean_f) ** 2 for x in S_list)
variance = variance_sum / len(S_list)
std_dev = variance ** 0.5

print("Variance:", variance)
print("Standard Deviation:", std_dev)

2+3+7+7+9
```

```python
def median(nums):
  nums.sort()
  n = len(nums)
  if n % 2 == 1:
    # For odd length, return the middle element
    return nums[n // 2]
  else:
    # For even length, return the average of the two middle elements
    return (nums[(n // 2) - 1] + nums[n // 2]) / 2

print("Median of [3, 1, 2]:", median([3, 1, 2]))
print("Median of [4, 1, 3, 2]:", median([4, 1, 3, 2]))
print("Median of [5]:", median([5]))
print("Median of [1, 2, 3, 4]:", median([1, 2, 3, 4]))
```
