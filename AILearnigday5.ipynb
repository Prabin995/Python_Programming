```python
def mean1(nums):
    mean1= sum(nums)/len(nums)
    return mean1
def std_dev(nums):
    m = mean1(nums)
    total = 0
    for x in nums:
        total = total + (x - m) ** 2    # वर्गको जोड
    variance = total / len(nums)          # वर्गहरूको औसत
    return variance ** 0.5               # वर्गमूल

print(std_dev([2, 4, 4, 4, 5, 5, 7, 9]))
print(std_dev([1, 9, 1, 9]))
print(std_dev([5, 5, 5]))
```

```python
emails = [
    {"lottery": True,  "unknown": True,  "link": True,  "bank": False},
    {"lottery": False, "unknown": True,  "link": True,  "bank": False},
    {"lottery": True,  "unknown": True,  "link": True,  "bank": True},
    {"lottery": False, "unknown": False, "link": False, "bank": False},
]

def check_spam(email):
    # यदि bank True छ भने सिधै "NOT_SPAM"
    if email["bank"]:
        return "NOT_SPAM"

    count = 0
    if email["lottery"]:
        count += 3
    if email["link"]:
        count += 1
    if email["unknown"]:
        count += 1

    if count >= 3:
        return "SPAM"
    else:
        return "NOT_SPAM"

# (क) P(spam) को गणना
spam = 0
for e in emails:
    result = check_spam(e)
    if result == "SPAM":
        spam = spam + 1

print("P(spam) =", spam / len(emails))

# (ख) lottery = True भएकाहरू मध्ये spam को अनुपात गणना
with_lottery = []
for e in emails:
    if e['lottery']:
        with_lottery.append(e)
print(with_lottery)

spam_l = 0
for e in with_lottery:
    result = check_spam(e)
    if result == "SPAM":
        spam_l = spam_l + 1

print("P(spam | lottery = True) =", spam_l / len(with_lottery))
```

```python
students = [
    {"name": "A", "marks": 70, "attends": True},
    {"name": "B", "marks": 30, "attends": True},
    {"name": "C", "marks": 55, "attends": False},
    {"name": "D", "marks": 20, "attends": False},
    {"name": "E", "marks": 45, "attends": True},
]

def is_pass(marks):
    return marks >= 40
passed = 0
for s in students:
    if is_pass(s["marks"]):
        passed = passed + 1

print(passed / len(students))

notattending = []                  # खाली list बाट सुरु
for s in students:
    if not s["attends"]:
        notattending.append(s)     # शर्त मिल्दा मात्र थप्ने

print(len(notattending))           # 3
passed_natt = 0
for s in notattending:             # सबै होइन, छुट्याएको list मा loop
    if is_pass(s["marks"]):
        passed_natt = passed_natt + 1

print(passed_natt / len(notattending))
```
