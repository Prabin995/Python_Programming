```python
#1.1 तीन तरिकाले import गर्ने
import math                 # तरिका 1: पूरा module
print(math.sqrt(16))        # 4.0   (module को नाम + . + function)

from math import sqrt       # तरिका 2: एउटा function मात्र
print(sqrt(16))             # 4.0   (नाम सिधै)

import math as m            # तरिका 3: छोटो उपनाम
print(m.sqrt(16))           # 4.0
```

```python
#1.2 math का उपयोगी function
print(math.sqrt(16))

print(math.floor(2.7))

print(math.floor(-2.5))

print(math.ceil(2.1))
print(math.ceil(-2.1))

print(math.pi)

print(math.factorial(5))
```

```python
#1.3 random (अनियमित सङ्ख्या)
import random


print(random.seed(42))                 # एउटै नतिजा दोहोर्‍याउन (बीज)
print(random.randint(1, 6))      # 1 देखि 6 सम्म, दुवै सहित (पासा)
print(random.choice(["A", "B", "C"]))   # list बाट एउटा छान्ने
nums = [1, 2, 3, 4, 5]
random.shuffle(nums)         # मूल list नै मिसाउँछ (None फर्काउँछ!)
print(random.shuffle(nums))
print(random.sample(nums, 2))    # नयाँ list, दुई वटा, दोहोर्‍याइँदैन
```

```python
#2.2 with प्रयोग गरेर लेख्ने
with open("scores.txt", "w", encoding="utf-8") as f:
    f.write("70\n")
    f.write("85\n")
    f.write("60\n")

```

```python
#2.3 पढ्ने तीन तरिका
with open("scores.txt", "r", encoding="utf-8") as f:
    text = f.read()        # सबै एकै string मा: "70\n85\n60\n"
print(repr(text))          # repr ले \n पनि देखाउँछ
```

```python
with open("scores.txt", "r", encoding="utf-8") as f:
    for line in f:         # एक-एक लाइन
        print(line.strip())  # strip() ले अगाडि-पछाडिको खाली ठाउँ र \n हटाउँछ
```

```python
with open("scores.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()  # list: ['70\n', '85\n', '60\n']
    print(lines)
```

```python
total = 0
with open("scores.txt", "r", encoding="utf-8") as f:
    for line in f:
        total = total + int(line.strip())   # int() ले सङ्ख्या बनाउँछ
print(total)    # 215
```

```python
#2.4 CSV: तालिका जस्तो file
#लेख्ने:
import csv

students = [
    {"name": "A", "marks": 70, "attends": True},
    {"name": "B", "marks": 30, "attends": True},
    {"name": "C", "marks": 55, "attends": False},
]

with open("students.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "marks", "attends"])
    writer.writeheader()          # पहिलो लाइन: शीर्षक
    for s in students:
        writer.writerow(s)
```

```python
#पढ्ने:
with open("students.csv", "r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```
