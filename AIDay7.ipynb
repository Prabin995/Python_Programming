```python
#📘 भाग 3: Error Handling (try / except)
#अहिलेसम्म error आउँदा program रोकिन्थ्यो। वास्तविक काममा गलत data आउँछ, तर पूरै program रोकिनु हुँदैन।
#3.1 आधारभूत ढाँचा

try:
    n = int("abc")           # यहाँ error आउँछ
    print("यो चल्दैन")
except ValueError:
    print("सङ्ख्या होइन")    # यो चल्छ
print("program चलिरह्यो")
```

```python
#3.3 धेरै except, else, finally
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("शून्यले भाग गर्न मिल्दैन")
        return None
    except TypeError:
        print("सङ्ख्या होइन")
        return None
    else:
        return result            # error आएन भने मात्र चल्छ

print(safe_divide(10, 2))     # 5.0
print(safe_divide(10, 0))     # छाप्छ, अनि None
print(safe_divide(10, "x"))   # छाप्छ, अनि None
```

```python
#3.4 raise: आफैँ error उठाउने
def check_marks(m):
    if m < 0 or m > 100:
        raise ValueError("अंक 0 देखि 100 बीच हुनुपर्छ")
    return m

try:
    check_marks(150)
except ValueError as e:
    print("गल्ती:", e)       # e मा सन्देश हुन्छ
```

```python
#🎓 Worked Example: सबै मिलाएर

#समस्या: grades.csv file मा विद्यार्थीका अंक छन्, केही अंक बिग्रेका छन्। उपस्थित भएकामा pass को अनुपात निकाल्ने।

#चरण 1: file बनाउने (बिग्रेको data सहित)

import csv

rows = [
    {"name": "A", "marks": "70",  "attends": "True"},
    {"name": "B", "marks": "abc", "attends": "True"},
    {"name": "C", "marks": "55",  "attends": "False"},
    {"name": "D", "marks": "45",  "attends": "True"},
]
with open("grades.csv", "w", newline="", encoding="utf-8") as f:
    w = csv.DictWriter(f, fieldnames=["name", "marks", "attends"])
    w.writeheader()
    w.writerows(rows)
```

```python
#चरण 2: पढ्ने र रूपान्तरण गर्ने (गलत अंक हटाउने)
def to_bool(text):
    return text == "True"

def load_students(filename):
    students = []
    skipped = 0
    try:
        with open(filename, "r", newline="", encoding="utf-8") as f:
            for row in csv.DictReader(f):
                try:
                    students.append({
                        "name": row["name"],
                        "marks": int(row["marks"]),
                        "attends": to_bool(row["attends"]),
                    })
                except ValueError:
                    skipped = skipped + 1
    except FileNotFoundError:
        print("file भेटिएन:", filename)
        return [], 0
    return students, skipped
```

```python
#चरण 3: पुरानो function पुन: प्रयोग गर्ने
def is_pass(marks):
    return marks >= 40

students, skipped = load_students("grades.csv")

attending = [s for s in students if s["attends"]]
passed = [s for s in attending if is_pass(s["marks"])]

print(len(students), skipped)             # 3 1
print(len(passed) / len(attending))       # 1.0
```
