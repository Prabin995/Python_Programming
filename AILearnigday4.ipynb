```python
def total(marks):
  total = 0
  for score in marks.values():
      total += score
  return total

marks = {"math": 70, "science": 85, "english": 60}
print(total(marks))

def biggetst(marks):
  highest_score = 0
  highest_subject = None
  for subject, score in marks.items():
    if score > highest_score:
      highest_score = score
      highest_subject = subject
  return highest_subject, highest_score

marks = {"math": 70, "science": 85, "english": 60}
subject, score = biggetst(marks)
print(f"{subject}: {score}")
```

```python
def higest(marks):
  big=0
  for score in marks.values():
    if score > big:
      big = score
  return big
marks = {"math": 70, "science": 85, "english": 60}
print(higest(marks))

def smallest(marks):
  small=100
  S_subjec=None
  for subject, score in marks.items():
    if score<small:
      small=score
      S_subjec=subject
  return S_subjec, small

marks = {"math": 70, "science": 85, "english": 60}
print(smallest(marks))

def precentage(marks):
  total=0
  for score in marks.values():
    total += score
  return total/len(marks)

marks = {"math": 70, "science": 85, "english": 60}
print(f"{precentage(marks)}%")

```

```python


def is_spam(emails):
  List = {"spam": 0, "non_spam": 0}
  for email in emails:
    if email["bank"]:
      List["non_spam"] += 1
      continue

    count = 0
    if email["lottery"]:
      count += 3
    if email["link"]:
      count += 1
    if email["unknown"]:
      count += 1

    if count >= 3:
      List["spam"] += 1
    else:
      List["non_spam"] += 1
  return List

emails = [
    {"lottery": True,  "unknown": True,  "link": True,  "bank": False},
    {"lottery": False, "unknown": True,  "link": True,  "bank": False},
    {"lottery": True,  "unknown": True,  "link": True,  "bank": True},
    {"lottery": False, "unknown": False, "link": False, "bank": False},]
print(is_spam(emails))
```
