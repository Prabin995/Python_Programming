```python
def find_biggest(nums):
    biggest = nums[0]
    for x in nums:
        if x > biggest:
            biggest = x
    return biggest

print(find_biggest([5, 1, 3, 0, 4, 2, 3, 1]))
print(find_biggest([9, 1, 1]))
```

```python
def check_spam(has_lottery_word, unknown_sender, has_link, sender_is_bank):
  negscore=0
  if  sender_is_bank:
    return "NOT SPAM"
  if has_lottery_word:
    negscore=negscore+3
  if unknown_sender:
    negscore=negscore+1
  if has_link:
    negscore =negscore+1
  if negscore>=3:
    return "SPAM"
  else:
    return "NOT SPAM"
a=check_spam(True,True,True,False)
b=check_spam(True,True,True,True)
c=check_spam(False,True,True,False)
d=check_spam(False,False,False,False)

print(a, b, c, d)
```

```python
def find_biggest(nums):
  biggest=nums[0]
  for x in nums:
    if  x>biggest:
      biggest=x
  return biggest

def find_smallest(nums):
    smallest=nums[0]
    for x in nums:
      if x<smallest:
        smallest=x
    return smallest

def spread(nums):
  a=find_biggest(nums)
  b=find_smallest(nums)
  return a-b

print(spread([5, 1, 3, 0, 4, 2, 3, 1]))
print(spread([7]))
print(spread([-5, -2, -9]))


```

```python
words = ["spam", "ham", "spam"]
counts = {}

for w in words:
    if w in counts:
        counts[w] = counts[w] + 1
    else:
        counts[w] = 1

print(counts)
```
