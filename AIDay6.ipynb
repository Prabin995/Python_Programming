```python
def median(nums):
    s = sorted(nums)          # नयाँ list, मूल सुरक्षित
    n = len(s)
    if n % 2 == 1:
        return s[n // 2]
    return (s[n // 2 - 1] + s[n // 2]) / 2
data = [3, 1, 2]; print(median(data)); print(data)
```

```python
#Q3 (🟡, भाग 1 + 2): तपाईंको median (Level 3.1) लाई यस्तो बनाउनुहोस् कि मूल list नबिग्रियोस्। परीक्षण: data = [3, 1, 2]; print(median(data)); print(data), अपेक्षित 2 र [3, 1, 2]। दुवै अपेक्षा पहिले हातले लेख्नुहोस्।
def median(data):
  s=sorted(data)
  n=len(s)
  if n%2==1:
    return s[n//2]
  return (s[n//2-1]+s[n//2])/2

data = [3, 1, 2]; print(median(data)); print(data)
```

```python
#Q4 (🟡, भाग 3): nums = [10, 20, 30, 40, 50, 60] मा हातले भन्नुहोस्: nums[1:4], nums[:2], nums[-2:], nums[::-1]। त्यसपछि scores = [5, 1, 3, 0, 4, 2, 3, 1] बाट सबैभन्दा ठूला तीन अंक निकाल्ने code लेख्नुहोस्।

nums = [10, 20, 30, 40, 50, 60]
print(nums[1:4])
print(nums[:2])
print(nums[-2:])
print(nums[::-1])

def max_nums(scores):
  Big= sorted(scores)[-3:]
  return Big
scores = [5, 1, 3, 0, 4, 2, 3, 1]
print(max_nums(scores))
```

```python
#Q5 (🔴, भाग 4 + 5): scores = [5, 1, 3, 0, 4, 2, 3, 1] मा:

#(क) list comprehension ले 3 वा बढी भएका अंकको list बनाउनुहोस्।
#(ख) त्यही सर्त पूरा गर्ने अंक कति वटा छन् भनेर निकाल्नुहोस् (सिधै len र comprehension प्रयोग गरेर)।
#(ग) min_max(nums) function बनाउनुहोस्, जसले सबैभन्दा सानो र सबैभन्दा ठूलो tuple को रूपमा फर्काओस् (Level 2.1 को find_smallest र find_biggest पुन: प्रयोग गरेर)। low, high = min_max(scores) लेखेर unpack गर्नुहोस्

scores = [5, 1, 3, 0, 4, 2, 3, 1]
#(क) list comprehension ले 3 वा बढी भएका अंकको list बनाउनुहोस्।
List_G3=[x for x in scores if x>=3]
print(List_G3)

#(ख) त्यही सर्त पूरा गर्ने अंक कति वटा छन् भनेर निकाल्नुहोस् (सिधै len र comprehension प्रयोग गरेर)।
count_G3= len(List_G3)
print(count_G3)

#(ग) min_max(nums) function बनाउनुहोस्, जसले सबैभन्दा सानो र सबैभन्दा ठूलो tuple को रूपमा फर्काओस् (Level 2.1 को find_smallest र find_biggest पुन: प्रयोग गरेर)। low, high = min_max(scores) लेखेर unpack गर्नुहोस्

def min_max(nums):
  return min(nums), max(nums)
low, high =min_max([5, 1, 3, 0, 4, 2, 3, 1])
print(low, high)
```
